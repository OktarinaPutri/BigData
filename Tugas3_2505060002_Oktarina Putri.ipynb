```python
# Sel ini menghasilkan TIGA dataset cabang (kota) terpisah untuk Tugas Mandiri Pertemuan 3
import numpy as np
import pandas as pd

kategori_list = ["Elektronik", "Fashion", "Makanan & Minuman", "Kesehatan & Kecantikan", "Rumah Tangga"]
metode_bayar_list = ["Transfer Bank", "E-Wallet", "COD", "Kartu Kredit"]
tanggal_range = pd.date_range("2026-08-01", "2026-08-31", freq="D")

cabang_kota = {"Magelang": 101, "Yogyakarta": 202, "Semarang": 303}

for kota, seed in cabang_kota.items():
    np.random.seed(seed)  # seed berbeda tiap kota agar datanya bervariasi, namun tetap konsisten/reproducible
    n = 200
    data_cabang = {
        "order_id": [f"{kota[:3].upper()}-{2000 + i}" for i in range(n)],
        "tanggal": np.random.choice(tanggal_range, size=n),
        "kategori": np.random.choice(kategori_list, size=n, p=[0.25, 0.25, 0.20, 0.15, 0.15]),
        "unit_terjual": np.random.randint(1, 8, size=n),
        "harga_satuan": np.random.choice([25000, 50000, 75000, 100000, 150000, 250000], size=n),
        "metode_pembayaran": np.random.choice(metode_bayar_list, size=n),
    }
    df_cabang = pd.DataFrame(data_cabang)
    df_cabang["kota"] = kota
    nama_file = f"transaksi_{kota.lower()}.csv"
    df_cabang.to_csv(nama_file, index=False)
    print(f"Berkas '{nama_file}' berhasil dibuat: {df_cabang.shape[0]} baris")

print("\nKetiga berkas CSV cabang siap digunakan untuk Tugas Mandiri.")
```

```python
import pandas as pd
df_magelang=pd.read_csv("magelang_dari_hdfs.csv")
df_semarang=pd.read_csv("semarang_dari_hdfs.csv")
df_yogyakarta=pd.read_csv("yogyakarta_dari_hdfs.csv")

df_gabungan=pd.concat([df_magelang,df_semarang,df_yogyakarta], ignore_index=True)
df_gabungan
```

```python
df_gabungan["kota"].value_counts()
```

```python
df_gabungan["total_pendapatan"] = df_gabungan["unit_terjual"] * df_gabungan["harga_satuan"]
df_gabungan.head()
```

```python
ringkasan_kota_kategori = df_gabungan.groupby(["kota","kategori"])["total_pendapatan"].sum().reset_index()
ringkasan_kota_kategori
```

```python
df_gabungan.to_csv("data_gabungan_bersih.csv", index=False)
ringkasan_kota_kategori.to_csv("ringkasan_kota_kategori.csv", index=False)
print("Kedua berkas berhasil disimpan didisk lokal.")
```

#Reflektif
Apa keuntungan menyimpan data mentah (raw) terpisah dari data olahan (processed) di HDFS, dibandingkan menyimpan semuanya bercampur dalam satu folder?

Dari praktikum yang sudah saya lakukan, menurut saya sendiri pemisahan folder raw dengan processed itu sangat berguna dan kita sebagai user mendapat keuntungan banyak dari prosesi ini. Dimana ketika kita memisahkan data mentah dengan raw, kita bisa menjaga sumber kebenaran. Dimana folder raw menyimpan data asli yang bersifat paten atau tidak boleh diubah. Jika terdapat kesalahan logika dalam proses pembersihan atau transformasi data, kita sebagai user dapat mengulang proses pengolahan dari data mentah awal tanpa risiko kehilangan data. Kemudian manfaat yang lain ialah mempermudah eksplorasi dan auditing. Dimana tim data akan dengan mudah mengetahui status kematangan data. Data di folder processed sudah siap dikonsumsi untuk analisis dan pembuatan laporan, sedangkan folder raw bisa digunakan jika membutuhkan variabel asli yang belum terfilter. Keuntungan selanjutnya adalah mencegah kerusakan data. Ini memfokuskan proses write dan update hanya pada folder processed menjaga folder raw tetap aman dari risiko terhapus secara tidak sengaja atau tertimpa oleh scitpt yang bermasalah. Dan keuntungan yang akan saya sebutkan terakhir ialah meningkatkan efisiensi dan keamanan operasional. Karenanya akses kontrol dpat diatur lebih ketat agar tidak membebani atau pembocoran data mentah.
